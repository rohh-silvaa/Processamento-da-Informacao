**3) Criação da função que usa a API para trazer o banco de notícias com base no termo de busca e a quantidade de matérias que gostaria que retornasse!**

```
def buscar_noticias(termo, quantidade):
    print(f"Buscando notícias sobre: {termo}...")

    # Busca notícias (endpoint 'everything' busca em todo o histórico disponível no plano free)
    response = newsapi.get_everything(
        q=termo,
        language='pt',
        sort_by='relevancy',
        page_size=quantidade
    )

    artigos = []
    if response['status'] == 'ok':
        for article in response['articles']:
            artigos.append({
                'titulo': article['title'],
                'descricao': article['description'] or "",
                'conteudo': article['content'] or "",
                'url': article['url']
            })
    else:
        print("Erro na busca:", response)

    return pd.DataFrame(artigos)
```
